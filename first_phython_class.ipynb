    # runs if all above were False

# You have 8000 in your wallet, simulate purchasing items worth 1200, 3500, 800, 5000, and 600 from an online store.
# Return "Insufficient funds" when your wallet balance is less than the item price and complete the purchase,
# if the item price is less than or equal to the wallet balance, while printing the remaining balance after each transaction.

````python
wallet_balance = 8000
````

````python
item_prices = [1200, 3500, 800, 5000, 600]
````

````python
for price in item_prices:
    if price <= wallet_balance:
        wallet_balance -= price
        print(f"Purchased item for {price}. Remaining balance: {wallet_balance}")
    else:
        print(f"Insufficient funds for item priced {price}. Remaining balance: {wallet_balance}")
````

````python
temps = [38, 22, 17, 40, 29, 15, 36]

hot_count = 0
warm_count = 0
cold_count = 0

for temp in temps:
    if temp >= 35:
        label = "Hot"
        hot_count += 1
    elif 20 <= temp <= 34:
        label = "Warm"
        warm_count += 1
    else:
        label = "Cold"
        cold_count += 1
    print(f"{temp}°C - {label}")

print("\nCounts:")
print(f"Hot days: {hot_count}")
print(f"Warm days: {warm_count}")
print(f"Cold days: {cold_count}")
````

````python
# Create the contacts dictionary with at least 4 people
contacts = {
    "John": "08012345678",
    "Mary": "08087654321",
    "Ngozi": "08011223344",
    "David": "08099887766"
}

# a) Print all names using a loop
print("Contacts:")
for name in contacts:
    print(name)

# b) Add a new contact
contacts["Amina"] = "08033445566"

# c) Remove one contact using pop()
removed_contact = contacts.pop("David", None)
print(f"\nRemoved contact: {removed_contact}")

# d) Check if "Ngozi" is in the contacts
if "Ngozi" in contacts:
    print("\nNgozi is in the contacts.")
else:
    print("\nNgozi is not in the contacts.")
````
